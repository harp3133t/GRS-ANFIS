# Download External Datasets

This notebook recreates the local `data/` snapshots used by the GRS-ANFIS reproducibility notebooks. The dataset files are intentionally not committed to GitHub.

Sources:

- Vowel: OpenML data id `307`; the OpenML record cites the UCI Vowel source.
- Spambase: UCI Machine Learning Repository id `94`.
- Breast Cancer Wisconsin (Original): UCI Machine Learning Repository id `15`.
- Gisette: UCI Machine Learning Repository id `170`; raw GISETTE files are downloaded from the UCI archive.

Run this notebook before `00_environment_and_data_audit.ipynb` on a fresh clone.

```python
from pathlib import Path
import subprocess
import sys


def find_project_root(start: Path) -> Path:
    start = start.resolve()
    markers = ["data.py", "model.py", "learning.py", "utils.py"]
    for path in [start, *start.parents]:
        if all((path / marker).exists() for marker in markers):
            return path
    raise RuntimeError("Could not find GRS-ANFIS project root")


PROJECT_ROOT = find_project_root(Path.cwd())
SCRIPT = PROJECT_ROOT / "scripts" / "download_datasets.py"
print(f"PROJECT_ROOT={PROJECT_ROOT}")
print(f"SCRIPT={SCRIPT}")
```

```python
# Add "--overwrite" to refresh existing local snapshots.
cmd = [sys.executable, str(SCRIPT)]
print(" ".join(cmd))
subprocess.run(cmd, cwd=PROJECT_ROOT, check=True)
```
